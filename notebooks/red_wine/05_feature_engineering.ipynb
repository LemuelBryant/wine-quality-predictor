## Purpose  
The purpose of this notebook is to prepare the predictor variables for machine learning by evaluating feature scaling and investigating whether engineered features improve model performance.

```python
# Load the cleaned dataset

import pandas as pd

df = pd.read_csv("../../data/processed/red_wine_cleaned.csv")
```

```python
display(df.head())
display(df.shape)
df.info()
```

```python
# Feature engineering

df["so2_ratio"] = (
    df["free sulfur dioxide"] /
    df["total sulfur dioxide"]
)

df["alcohol_density_ratio"] = (
    df["alcohol"] /
    df["density"]
)

df.head()
```

```python
df.to_csv("../../data/processed/red_wine_features.csv", index=False)
```
