```python
import pandas as pd

df = pd.read_csv("../../data/raw/winequality-white.csv", sep=";")

df.head()
```

```python

duplicates = int(df.duplicated().sum())
print(f"The number of duplicates is {duplicates}.")
```

## Duplicate Records

The dataset contains **937 duplicate rows**.

These duplicate observations represent approximately **19%** of the dataset. Duplicate records may affect model training by causing certain examples to be weighted more heavily than others, potentially influencing model performance and generalization.

Before model training, duplicate records should be investigated further to determine whether they represent:

* True repeated measurements of the same wine sample
* Data collection artifacts
* Valid observations that should remain in the dataset

```python
duplicates = df[df.duplicated(keep=False)].sort_values(by=list(df.columns))
duplicates.head(10)
```

## Duplicate Record Investigation

After examining the duplicate rows, many duplicates appear at very close index locations. For example, several identical records occur in adjacent or nearly adjacent positions within the dataset. This pattern suggests that the duplicates were likely introduced during the data collection, compilation, or processing stages rather than representing independent wine samples.

Because these duplicate rows contain identical values across all physicochemical measurements and the target variable (`quality`), they do not provide additional information for model training. Keeping them in the dataset could cause certain observations to receive more weight during model fitting and potentially bias model performance.

Therefore, duplicate records will be removed.


```python
df = df.drop_duplicates()
print(f"Dataset shape after removing duplicates: {df.shape}")
print(f"Remaining duplicates: {df.duplicated().sum()}")
```

## Save the Cleaned Dataset

```python
df.to_csv("../../data/processed/white_wine_cleaned.csv", index=False)
```
